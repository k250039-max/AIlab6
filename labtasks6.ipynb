## Task 1:

```python
import numpy as np

marks = np.array([
    [78, 65, 82, 70, 88],
    [45, 52, 48, 55, 50],
    [90, 85, 92, 88, 95],
    [60, 72, 68, 75, 64],
    [35, 42, 50, 38, 45],
    [80, 76, 85, 79, 82],
    [55, 61, 58, 64, 60],
    [95, 91, 89, 94, 90],
    [48, 45, 52, 49, 51],
    [70, 68, 75, 72, 78]
])

totals = np.sum(marks, axis=1)
avgs = np.mean(marks, axis=1)

avg_subjects = np.mean(marks, axis=0)

max_avg = np.max(avgs)
student = np.argmax(avgs) + 1

count = np.sum(avgs >= 50)

status = np.where(avgs >= 50, "PASS", "FAIL")

print("Student totals:", totals)
print("Student averages:", avgs)
print("Subject averages:", avg_subjects)
print(f"Student {student} has Max average: {max_avg}")
print("Students with average >= 50:", count)
print("Status:", status)
```

## Task 02:

```python
import numpy as np

X = np.array([
    [20, 50000, 5, 80],
    [25, 60000, 7, 75],
    [22, 55000, 6, 85],
    [30, 70000, 8, 90],
    [28, 65000, 7, 88],
    [24, 52000, 5, 78],
    [35, 80000, 10, 95],
    [27, 62000, 6, 82],
    [32, 75000, 9, 92],
    [21, 48000, 4, 70]
])

means = np.mean(X,axis=0)
print(f"Means:{means}")
stds = np.std(X,axis=0)
print(f"Standard deviations:{stds}")

Standardized = (X-means)/stds

print(f"Standardized Matrix:\n{Standardized}")
print(f"New means:{np.mean(Standardized,axis=0)}")
print(f"New Deviations:{np.std(Standardized,axis=0)}")
```

## Task 04


```python
import numpy as np

readings = np.array([
    [45, 52, 48, 55, 60, 58],
    [70, 65, 72, 68, 75, 80],
    [30, 35, 32, 40, 38, 42],
    [85, 90, 88, 92, 95, 91],
    [55, 50, 58, 62, 60, 65]
]) 
sensor_mean = np.round(np.mean(readings, axis=1),2)
sensor_min = np.min(readings, axis=1)
sensor_max = np.max(readings, axis=1)
sensor_std = np.round(np.std(readings, axis=1),2)

highest_sensor = np.argmax(sensor_mean) + 1

threshold = 70
above_threshold = readings > threshold
print("Sensor Mean:", sensor_mean)
print("Sensor Minimum:", sensor_min)
print("Sensor Maximum:", sensor_max)
print("Sensor Standard Deviation:", sensor_std)

print("Sensor with Highest Average Reading:", highest_sensor)

print("Readings Above Threshold:")
print(above_threshold)
```

## Task 05:

```python
import numpy as np

A = np.array([
    [4, 2],
    [1, 3]
])

B = np.array([
    [2, 5],
    [6, 1]
])

print("Shape of A:", A.shape)
print("Shape of B:", B.shape)

print("\nAddition:")
print(A + B)

print("\nSubtraction:")
print(A - B)

print("\nElement-wise Multiplication:")
print(A * B)

print("\nMatrix Multiplication:")
print(A @ B)

print("\nTranspose of A:")
print(A.T)

print("\nTranspose of B:")
print(B.T)

print("\nDeterminant of A:")
print(np.linalg.det(A))

print("\nDeterminant of B:")
print(np.linalg.det(B))

print("\nInverse of A:")
print(np.linalg.inv(A))

print("\nInverse of B:")
print(np.linalg.inv(B))
```

## Task 06

```python
import numpy as np

rng1 = np.random.default_rng(42)
data1 = rng1.normal(loc=50, scale=5, size=1000)

mean1 = data1.mean()
median1 = np.median(data1)
std1 = data1.std()
min1 = data1.min()
max1 = data1.max()
within1 = np.mean(np.abs(data1 - mean1) <= std1) * 100

print("Seed 42")
print("Mean:", mean1)
print("Median:", median1)
print("Std Dev:", std1)
print("Min:", min1)
print("Max:", max1)
print("% within 1 SD:", within1)

rng2 = np.random.default_rng(123)
data2 = rng2.normal(loc=50, scale=5, size=1000)

mean2 = data2.mean()
median2 = np.median(data2)
std2 = data2.std()
min2 = data2.min()
max2 = data2.max()
within2 = np.mean(np.abs(data2 - mean2) <= std2) * 100

print("\nSeed 123")
print("Mean:", mean2)
print("Median:", median2)
print("Std Dev:", std2)
print("Min:", min2)
print("Max:", max2)
print("% within 1 SD:", within2)


print("\nDifference (Seed 123 - Seed 42)")
print("Mean:", mean2 - mean1)
print("Median:", median2 - median1)
print("Std Dev:", std2 - std1)
print("Min:", min2 - min1)
print("Max:", max2 - max1)
print("% within 1 SD:", within2 - within1)
```

## Task 07:

```python
import numpy as np

arr = np.array([12.5, 7.0, np.nan, 20.0, 15.5, np.nan, 9.0, 18.0, np.nan, 11.0])
print("Original array:", arr)

nan_mask = np.isnan(arr)
print("NaN mask:", nan_mask)
print("Number of NaNs:", np.sum(nan_mask))
print("NaN positions:", np.where(nan_mask)[0])

print("Mean:", np.nanmean(arr))
print("Median:", np.nanmedian(arr))
print("Std Dev:", np.nanstd(arr))
print("Min:", np.nanmin(arr))
print("Max:", np.nanmax(arr))
print("Sum:", np.nansum(arr))
print("Regular np.mean:", np.mean(arr))

mean_val = np.nanmean(arr)
filled = np.where(np.isnan(arr), mean_val, arr)
print("Filled array:", filled)

print("Any NaN left?", np.isnan(filled).any())
print("NaN count after filling:", np.sum(np.isnan(filled)))
```
